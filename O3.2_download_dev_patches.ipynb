```python
import csv
import os
import urllib.parse
import urllib.request
```

```python
def download_pr_patches(csv_file_path):
    # Create the output directory if it does not exist
    output_dir = "developer_fixes"
    os.makedirs(output_dir, exist_ok=True)

    # Open and read the CSV file
    with open(csv_file_path, mode="r", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            pr_link = row["PR Link"].strip()
            pr_id=row["PR Id"].strip()
            # Skip empty rows
            if not pr_link:
                continue

            # Append .diff to the URL
            diff_url = f"{pr_link}.diff"

            # Create a safe filename by encoding the URL characters

            safe_filename = "PR_" + urllib.parse.quote(pr_id, safe="") + ".patch"
            file_path = os.path.join(output_dir, safe_filename)

            print(f"Downloading: {diff_url}")

            try:
                # Download the diff content and save it as a .patch file
                urllib.request.urlretrieve(diff_url, file_path)
                print(f"Saved to: {file_path}")
            except Exception as e:
                print(f"Failed to download {diff_url}. Error: {e}")



```

```python
# Example usage:
# Replace 'prs.csv' with your actual CSV file path
# Replace 'pr_link' with the exact name of your PR column
download_pr_patches("sampled_prs.csv")
```
